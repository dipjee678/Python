## Operator

```python
x =10 ; y = 3

sum = x + y
sub = x - y
mult = x*y
div = round(x/y,2)

rem = x % y

print(sum,sub,mult,div,rem)
```

## Comparison Operator


```python
x =10 ; y = 3


greater_then = x > y
greater_then_or_equal = 10 >= 10

print(greater_then,greater_then_or_equal)


less_then = x < y
less_then_or_equal = 10 <= 10

print(less_then,less_then_or_equal)


equal = x==10

print(equal)

not_equal = x != y

print(not_equal)
```

## Logical Operator
