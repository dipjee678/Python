## Problem : Digit Summation

```python
x = int(input())

y = int(input())

last_digit_of_x = x % 10
last_digit_of_y = y % 10

sum = last_digit_of_x + last_digit_of_y

print(sum)
```

## Max And Min

```python
inp = input()

numbers = inp.split()

num_1 = int(numbers[0])

num_2 = int(numbers[1])

num_3 = int(numbers[2])

min = num_1

max = num_1

if  num_2 < min:
    min = num_2

if  num_3 < min:
    min = num_3


if  num_2 > max:
    max = num_2

if  num_3 > max:
    max = num_3

print(min,max)
```

## Even , Odd , Positive , Negative

```python
n = int(input())

inp = input()

numbers = inp.split()

Even = 0 ; Odd= 0 ; Positive= 0; Negative = 0

for i in range(n):
    x = int(numbers[i])

    if x > 0:
        Positive+=1
    elif x < 0:
        Negative+=1

    if x % 2 !=0:
        Odd+=1
    else:
        Even+=1

print("Even:",Even)
print("Odd:",Odd)
print("Positive:",Positive)
print("Negative:",Negative)





```
