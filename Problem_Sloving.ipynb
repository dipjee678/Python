## Problem : Digit Summation

```python
x = int(input())

y = int(input())

last_digit_of_x = x % 10
last_digit_of_y = y % 10

sum = last_digit_of_x + last_digit_of_y

print(sum)
```

```python
inp = input()

numbers = inp.split()

num_1 = int(numbers[0])

num_2 = int(numbers[1])

last_digit_of_num_1 = num_1 % 10
last_digit_of_num_2 = num_2 % 10

sum = last_digit_of_num_1 + last_digit_of_num_2

print(sum)
```
