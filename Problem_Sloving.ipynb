## Max And Min

```python
inp = input()

numbers = inp.split()

num_1 = int(numbers[0])

num_2 = int(numbers[1])

num_3 = int(numbers[2])

min = num_1

max = num_1

if  num_2 < min:
    min = num_2

if  num_3 < min:
    min = num_3


if  num_2 > max:
    max = num_2

if  num_3 > max:
    max = num_3

print(min,max)
```

## Even , Odd , Positive , Negative

```python
n = int(input())

inp = input()

numbers = inp.split()

Even = 0 ; Odd= 0 ; Positive= 0; Negative = 0

for i in range(n):
    x = int(numbers[i])

    if x > 0:
        Positive+=1
    elif x < 0:
        Negative+=1

    if x % 2 !=0:
        Odd+=1
    else:
        Even+=1

print("Even:",Even)
print("Odd:",Odd)
print("Positive:",Positive)
print("Negative:",Negative)





```

## Sensor Calibrator — Min-Max Scaling

```python
inp = input()

numbers = inp.split()

x = float(numbers[0])

min_v = float(numbers[1])

max_v = float(numbers[2])

norm = (x - min_v) / (max_v - min_v)

print(f"{norm:.2f}")

```

## Majority Vote Gate

```python
n = int(input())
count_Yes=0
count_No=0
for i in range(n):
    target = input()
    if target == "YES":
        count_Yes=count_Yes+1
    elif target == "NO":
        count_No+=1

if count_Yes >= count_No:
    print("ACCEPT")
else:
    print("REJECT")


```

## AI Brightness Analyzer

```python
inp = input()
numbers = inp.split()

length = len(numbers)
total = 0

for i in range(length):
    pixel = int(numbers[i])   
    total += pixel

average_brightness = total / length

if average_brightness < 85:
    print("Dark Image")
elif 85 <= average_brightness <= 170:
    print("Normal Image")
else:
    print("Bright Image")
```

## AI Bias Detector

```python
inp = input()
numbers = inp.split()

length = len(numbers)
count_A=0
count_B=0
total_count=0

for i in range(length):
    pixel = numbers[i]   
    total_count=count_A+count_B
    if pixel == "A":
        count_A+=1
    elif pixel == "B":
        count_B+=1


percentage_A=(count_A/total_count)*100
percentage_B=(count_B/total_count)*100


if percentage_A>=70 or percentage_B>=70:
    print("Biased Model")
else:
    print("Fair Model")

```
