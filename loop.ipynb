## For Loop

```python
for i in range(1,11,2):
    print(i,"Hello World")
```

```python
for i in range(1,10):
    for j in range(1,5):
        print(i,j)
```

## While Loop

```python
count = 1

while count < 10 :
    print("Hello World")
    count = count + 1
```

## Break and Continue Statement

```python
sum = 0

for i in range(1 ,10):
    print(i,"is processing")
    if i%2==0:
        continue
    sum+=i
    print(i,"is added to the sum")
    
```

```python
accuracy = 30

for i in range(100):
    accuracy+=1
    print(accuracy)
    if accuracy == 100:
        break
```
