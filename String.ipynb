## String Methods

```python
string = "Welcome to Phitron ML Course. Phitron"

processed_string = string.lower()


print(len(string))

print("phitron" in processed_string)


index = processed_string.find("phitron")
print(index)

rindex = processed_string.rfind("phitron")
print(rindex)


count = processed_string.count("phitron")
print(count)

new_string = string.replace("ML", "AI/ML")
print(new_string)
```

## String Spliting, Joining and Formating

```python
prompt = "What is Phitron ?"

tokens = prompt.split()
print(tokens)
print(type(tokens))

sentence = '-'.join(tokens)
print(sentence)
print(type(sentence))
```

## List Comprehension

```python
even = []

for i in range (100):
    if i%2==0:
        even.append(i)
print(even)

```

```python
list = [x for x in range(100) if x%2==0]
print(list)
```

```python
fruits = ['apple' , 'orange' , 'lichi']

capitalize_fruits=[]

for fruit in fruits:
    capitalize_fruits.append(fruit.capitalize())

print(capitalize_fruits)
```

```python
upper_fruits=[fruit.upper() for fruit in fruits ]

print(upper_fruits)
```
